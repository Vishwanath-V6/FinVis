# FAISS-backed inference demo
This notebook requires a locally indexed PDF and downloads the QA model on first use.


```python
from src.retrieval import RetrievalIndex
from src.inference import load_qa_model, answer_question

index = RetrievalIndex.load("models/saved_model")
tokenizer, model = load_qa_model()
question = "What was the reported revenue?"
result = answer_question(question, index, tokenizer, model)
print(result["answer"])
print(result["sources"])
for evidence in result["evidence"]:
    print(evidence["source_filename"], evidence["page_number"], evidence["chunk_id"])
    print(evidence["text"])
```
